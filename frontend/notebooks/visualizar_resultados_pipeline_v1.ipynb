```
import sqlite3
import pandas as pd
from IPython.display import display, HTML
from pathlib import Path

db_path = Path("../../data/sjur_recortes.db")
assert db_path.exists(), f"❌ Banco de dados não encontrado: {db_path}"

def carregar_dataframe(query):
    conn = sqlite3.connect(db_path)
    df = pd.read_sql_query(query, conn)
    conn.close()
    return df
```

```
df_emails = carregar_dataframe("SELECT id, message_id, data_processamento FROM emails_processados ORDER BY data_processamento DESC")
display(HTML("<h3>📧 Emails Processados</h3>"))
display(df_emails)
```

```
# Escolha um ID da tabela acima
email_id_escolhido = df_emails.iloc[0]['id'] if not df_emails.empty else None
print(f"🔍 Email selecionado: ID={email_id_escolhido}")
```

```
if email_id_escolhido is not None:
    message_id = df_emails[df_emails['id'] == email_id_escolhido]['message_id'].values[0]
    display(HTML(f"<hr><h4>📌 Detalhamento para Message ID: {message_id}</h4>"))

    # Classificação (publicações)
    df_pub = carregar_dataframe(f"SELECT texto, tipo FROM publicacoes WHERE message_id = '{message_id}'")
    if not df_pub.empty:
        display(HTML("<b>📎 Classificação da Publicação:</b>"))
        display(df_pub)
    else:
        display(HTML("<i>⚠️ Nenhuma publicação encontrada.</i>"))

    # Partes
    df_partes = carregar_dataframe(f"SELECT parte, papel FROM partes WHERE message_id = '{message_id}'")
    if not df_partes.empty:
        display(HTML("<b>👤 Partes Extraídas:</b>"))
        display(df_partes)
    else:
        display(HTML("<i>⚠️ Nenhuma parte encontrada.</i>"))

    # Metadados
    df_meta = carregar_dataframe(f"SELECT chave, valor FROM metadados WHERE message_id = '{message_id}'")
    if not df_meta.empty:
        display(HTML("<b>🧾 Metadados:</b>"))
        display(df_meta)
    else:
        display(HTML("<i>⚠️ Nenhum metadado encontrado.</i>"))
else:
    display(HTML("⚠️ Nenhum email processado disponível para detalhar."))
```
