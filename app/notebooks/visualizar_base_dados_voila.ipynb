```
import sqlite3
import pandas as pd
from IPython.display import display, HTML
from pathlib import Path

# Caminho robusto para o banco de dados
db_path = Path("../../data") / "sjur_recortes.db"

def carregar_dataframe(query):
    conn = sqlite3.connect(db_path)
    df = pd.read_sql_query(query, conn)
    conn.close()
    return df
```

```
# Tabela mestre
df_mestre = carregar_dataframe("""
    SELECT id, message_id, data_processamento 
    FROM emails_processados 
    ORDER BY data_processamento DESC
""")
display(HTML("<h3>📧 Emails Processados</h3>"))
display(df_mestre)
```

```
# Tabelas detalhes para cada email
for index, row in df_mestre.iterrows():
    email_id = row["id"]
    display(HTML(f"<hr><h4>📝 Detalhes do Email ID {email_id}</h4>"))

    # HTML bruto
    df_html = carregar_dataframe(f"""SELECT conteudo_html FROM emails_html WHERE email_id = {email_id}""")
    if not df_html.empty:
        display(HTML("<b>Conteúdo HTML:</b>"))
        display(HTML(df_html.iloc[0]['conteudo_html']))
    else:
        display(HTML("<i>Sem conteúdo HTML registrado.</i>"))

    # JSON processado
    df_json = carregar_dataframe(f"""SELECT conteudo_json FROM emails_json WHERE email_id = {email_id}""")
    if not df_json.empty:
        display(HTML("<b>Conteúdo JSON:</b>"))
        display(HTML(f"<pre>{df_json.iloc[0]['conteudo_json']}</pre>"))
    else:
        display(HTML("<i>Sem conteúdo JSON registrado.</i>"))
```
